# SafeNet - Exploratory Data Analysis & Model Evaluation

This notebook documents the exploratory data analysis (EDA), feature correlation studies, and model evaluation benchmarks for the **SafeNet** AI-powered digital wellbeing risk prediction pipeline.

```python
import pandas as pd
import numpy as np
import json

# Load generated synthetic dataset
df = pd.read_csv('../data/digital_wellbeing_dataset.csv')
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Class distribution
print("Risk Distribution:\n", df['risk_level'].value_counts())
df.groupby('risk_level')[['daily_screen_time_hours', 'avg_session_mins', 'night_time_mins', 'social_media_hours']].mean()
```

```python
# Inspect Model Evaluation Metrics
with open('../models/training_report.json') as f:
    report = json.load(f)
print("Best Selected Model:", report['best_model'])
print("Feature Importances / Weights:", json.dumps(report['feature_importance'], indent=2))
```
